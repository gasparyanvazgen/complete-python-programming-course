## 1. Strings: a quick review

A **string** (`str`) is a sequence of characters. Strings are written inside single quotes, double quotes, or triple quotes.

````python
first_name = "Ada"
message = 'Python is fun!'
multiline_text = """This is a text
that uses more than one line."""

print(first_name)
print(message)
print(multiline_text)
print(type(message))
````

### Escape characters

Some characters have a special meaning inside strings:

| Character | Meaning |
|---|---|
| `\n` | New line |
| `\t` | Tab |
| `\'` | Single quote |
| `\"` | Double quote |
| `\\` | Backslash |

A raw string, written with an `r` prefix, treats backslashes literally. It is often useful for file paths and regular expressions.

````python
print("First line\nSecond line")
print("Name:\tAda")
print('It\'s Python.')

path = r"C:\Users\Ada\Documents"
print(path)
````

### Strings are immutable

Strings cannot be changed in place. A string method returns a new string; it does not modify the original string.

````python
word = "python"
upper_word = word.upper()

print(word)        # The original string is unchanged.
print(upper_word)  # A new string is returned.
````

## 2. Indexing: getting one character

Every character has a position called an **index**. Positive indexes start at `0`. Negative indexes count from the end, starting at `-1`.

For the string `"PYTHON"`:

| Character | P | Y | T | H | O | N |
|---|---:|---:|---:|---:|---:|---:|
| Positive index | 0 | 1 | 2 | 3 | 4 | 5 |
| Negative index | -6 | -5 | -4 | -3 | -2 | -1 |

````python
language = "PYTHON"

print(language[0])    # P
print(language[2])    # T
print(language[-1])   # N
print(language[-2])   # O
````

An index outside the valid range causes an `IndexError`.

````python
language = "PYTHON"
print(language[6])  # Try running this line and read the error.
````

## 3. Slicing: extracting part of a string

The general syntax is:

```python
text[start:stop:step]
```

- `start`: the index where the slice begins (included);
- `stop`: the index where the slice ends (excluded);
- `step`: how many positions to move each time.

You may omit one or more values. The default step is `1`.

````python
text = "PYTHON PROGRAMMING"

print(text[0:6])    # PYTHON
print(text[:6])     # Same result: start from the beginning
print(text[7:])     # PROGRAMMING
print(text[:8])     # First 8 characters
print(text[10:])    # From index 10 to the end
print(text[::3])    # Every third character
print(text[::-1])   # The whole string in reverse
print(text[6:4:-1]) # Move backwards: indexes 6 and 5
````

### Understand each slice

For `text = "ABCDEFGHIJKL"`:

| Expression | Meaning |
|---|---|
| `text[0:3]` | indexes 0, 1, 2 → `ABC` |
| `text[:8]` | from the beginning up to (not including) index 8 |
| `text[10:]` | from index 10 to the end |
| `text[::3]` | every third character |
| `text[::-1]` | reverse the string |
| `text[6:4:-1]` | start at index 6, move backwards, stop before index 4 |
| `text[4:4]` | an empty string because start and stop are equal |

A positive step moves from left to right. A negative step moves from right to left. The stop position is always excluded.

````python
text = "ABCDEFGHIJKL"

print(text[0:3])
print(text[:8])
print(text[10:])
print(text[::3])
print(text[6:4:-1])
print(text[4:4])
````

## 4. Joining and repeating strings

The `+` operator joins strings. The `*` operator repeats a string. The operands of `+` must both be strings.

````python
first = "Hello"
second = "Python"

print(first + " " + second)
print("ha" * 3)
print("-" * 20)

age = 21
# print("Age: " + age)  # This raises TypeError.
print("Age: " + str(age))
````

## 5. Membership: `in` and `not in`

Use `in` to check whether a substring exists inside another string. The check is **case-sensitive**.

````python
sentence = "this is a text in english"

print("a text" in sentence)       # True
print("text" in sentence)          # True
print("Python" in sentence)        # False
print("english" not in sentence)   # False
print("English" in sentence)       # False: capitalization differs
````

## 6. String methods: inspect and transform text

A **method** is a function that belongs to an object. String methods are called with dot notation, such as `text.lower()`.

### 6.1 Case conversion

| Method | What it does |
|---|---|
| `lower()` | converts letters to lowercase |
| `upper()` | converts letters to uppercase |
| `capitalize()` | capitalizes the first character and lowercases the rest |
| `title()` | capitalizes the beginning of words |
| `swapcase()` | changes uppercase to lowercase and vice versa |
| `casefold()` | performs stronger normalization for case-insensitive comparisons |

````python
text = "pYtHoN programming IS FUN"

print(text.lower())
print(text.upper())
print(text.capitalize())
print(text.title())
print(text.swapcase())
````

### 6.2 Removing whitespace and selected characters

- `strip()` removes matching characters from both ends.
- `lstrip()` removes them from the left.
- `rstrip()` removes them from the right.

Without an argument, these methods remove whitespace (spaces, tabs, and new lines). They do not remove characters from the middle of a string.

````python
user_input = "   Ada Lovelace   \n"

print(repr(user_input))
print(repr(user_input.strip()))
print(repr(user_input.lstrip()))
print(repr(user_input.rstrip()))

code = "---PYTHON---"
print(code.strip("-"))
````

### 6.3 Searching and counting

| Method / operator | Purpose |
|---|---|
| `count(sub)` | counts non-overlapping occurrences |
| `find(sub)` | returns the first matching index, or `-1` if not found |
| `index(sub)` | returns the first matching index, but raises `ValueError` if not found |
| `rfind(sub)` | searches from the right; returns `-1` if not found |
| `rindex(sub)` | searches from the right; raises `ValueError` if not found |
| `startswith(prefix)` | checks the beginning of a string |
| `endswith(suffix)` | checks the end of a string |
| `sub in text` | checks whether a substring exists |

`index()` and `find()` can accept optional `start` and `end` positions. The `end` position is excluded, just like slicing.

````python
text = "banana bread"

print(text.count("a"))
print(text.find("na"))
print(text.find("xy"))       # -1: not found
print(text.index("na"))
print(text.index("a", 2, 5))  # Search only in the selected range
print(text.startswith("ban"))
print(text.endswith("bread"))
print(text.rfind("a"))

# Compare find() and index():
print("cat".find("z"))
# print("cat".index("z"))  # Uncomment to see ValueError.
````

### 6.4 Replacing text

`replace(old, new, count)` returns a new string in which matching parts are replaced. The optional `count` limits the number of replacements.

````python
message = "I like Java. Java is popular."

print(message.replace("Java", "Python"))
print(message.replace("Java", "Python", 1))
print(message)  # The original is unchanged.
````

### 6.5 Splitting and joining

- `split(separator)` divides a string and returns a list.
- Without a separator, `split()` treats consecutive whitespace as one separator and ignores leading/trailing whitespace.
- `rsplit()` splits starting from the right.
- `splitlines()` splits at line boundaries.
- `separator.join(iterable)` joins strings using the separator.
- `partition(separator)` splits into at most three parts: before, separator, after.

````python
full_name = "Ada Lovelace Byron"
parts = full_name.split()
print(parts)

csv_line = "red,green,blue"
colors = csv_line.split(",")
print(colors)

path = "home/user/projects/app.py"
print(path.rsplit("/", 1))

lines = "first line\nsecond line\nthird line"
print(lines.splitlines())

print(" | ".join(colors))
print("user@example.com".partition("@"))
````

### 6.6 Checking what a string contains

These methods return `True` or `False`. Most of them return `False` for an empty string.

| Method | Returns `True` when... |
|---|---|
| `isalpha()` | all characters are letters |
| `isdigit()` | all characters are digits |
| `isdecimal()` | all characters are decimal characters |
| `isnumeric()` | all characters are numeric characters |
| `isalnum()` | all characters are letters or numbers |
| `isspace()` | all characters are whitespace |
| `islower()` | all cased letters are lowercase |
| `isupper()` | all cased letters are uppercase |
| `istitle()` | the string follows title-case rules |

````python
print("Python".isalpha())
print("12345".isdigit())
print("Python3".isalnum())
print("   ".isspace())
print("hello".islower())
print("HELLO".isupper())
print("Hello World".istitle())
````

### 6.7 Alignment and padding

| Method | Purpose |
|---|---|
| `center(width)` | centers text |
| `ljust(width)` | aligns text to the left |
| `rjust(width)` | aligns text to the right |
| `zfill(width)` | pads the left side with zeroes |
| `removeprefix(prefix)` | removes a matching prefix |
| `removesuffix(suffix)` | removes a matching suffix |

````python
label = "Python"

print(label.center(20, "-"))
print(label.ljust(12, "."))
print(label.rjust(12, "."))
print("42".zfill(5))
print("https://example.com".removeprefix("https://"))
print("report.csv".removesuffix(".csv"))
print("@username".removeprefix("@"))
print("Mr. Smith".removeprefix("Mr. "))
````

- `type(value)` shows the object's type.
- `dir(value)` lists its available attributes and methods.
- `help(value)` displays documentation.
- `help(str.find)` displays help for one method.

Names beginning and ending with double underscores (for example, `__add__`) are special methods. They are useful to recognize, but you do not need to study all of them now.

## 8. f-strings and format specifiers

An f-string lets you place expressions inside `{}`. Put `f` before the opening quote.

````python
name = "Ada"
age = 36
print(f"My name is {name} and I am {age} years old.")
print(f"Next year I will be {age + 1}.")
````

### 8.1 Number formatting

| Format specifier | Meaning |
|---|---|
| `:.2f` | floating-point number with 2 digits after the decimal point |
| `:d` | decimal integer |
| `:,` | thousands separator |
| `:b` | binary representation of an integer |
| `:x` | hexadecimal representation of an integer |
| `:.1%` | percentage with 1 digit after the decimal point |

````python
price = 19.9876
quantity = 3
number = 42
ratio = 0.8567
large_number = 1250000

print(f"Price: {price:.2f}")
print(f"Quantity: {quantity:d}")
print(f"Large number: {large_number:,}")
print(f"Binary: {number:b}")
print(f"Hexadecimal: {number:x}")
print(f"Success rate: {ratio:.1%}")
````

### 8.2 Alignment, width, and fill characters

- `>` aligns to the right.
- `<` aligns to the left.
- `^` centers the value.
- A number before the alignment symbol specifies the total width.
- A fill character can be placed before the alignment symbol.

````python
word = "Python"

print(f"|{word:>12}|")
print(f"|{word:<12}|")
print(f"|{word:^12}|")
print(f"|{word:*^12}|")
print(f"|{42:0>8}|")
print(f"|{42:0<8}|")
````

### Checking types with `isinstance()`

Use `isinstance(object, classinfo)` to check whether an object belongs to a type.

- `object` is the value you want to check.
- `classinfo` is a class/type, or a tuple of types.
- The function returns `True` or `False`.

Prefer `isinstance()` when checking a value’s type because it also recognizes subclasses. For example, `bool` is a subclass of `int`, so `isinstance(True, int)` returns `True`.


````python
age = 21
price = 19.99
name = "Ani"

print(isinstance(age, int))                 # True
print(isinstance(price, (int, float)))     # True
print(isinstance(name, str))               # True
print(isinstance(price, int))              # False
print(isinstance(True, int))               # True: bool is a subclass of int
````

### `min()` and `max()` with strings

When strings are compared, Python uses lexicographic (Unicode code point) order. This is not necessarily alphabetical order in the way a human dictionary treats uppercase and lowercase letters. For a case-insensitive comparison, use `key=str.lower`.

````python
names = ["zoe", "Ada", "bob", "Charlie"]

print(min(names))
print(max(names))
print(min(names, key=str.lower))
print(max(names, key=str.lower))
````

### `enumerate()` and `zip()`

````python
students = ["Ani", "David", "Mariam"]

for index, student in enumerate(students, start=1):
    print(index, student)

names = ["Ani", "David", "Mariam"]
points = [90, 85, 97]

for student, point in zip(names, points):
    print(f"{student}: {point}")
````

## 11. Dates and times with `datetime`

Python's `datetime` module provides classes for working with dates and times.

| Class | Stores |
|---|---|
| `date` | year, month, day |
| `time` | hour, minute, second, microsecond |
| `datetime` | date and time together |

Import the classes you need from `datetime`.

````python
from datetime import date, time, datetime

print(date.today())
print(time(14, 30, 0))
print(datetime.now())
````

### 11.1 Format a date and time with `strftime()`

`strftime()` means **string format time**. It converts a `date` or `datetime` object into a readable string using format codes.

The example below uses a fixed date and time so that everyone sees the same output when running the notebook:

- `today.strftime("%d.%m.%Y")` produces a numeric date such as `10.11.2026`.
- `today.strftime("%B %d, %Y")` uses the full month name, such as `November 10, 2026`.
- `today.strftime("%A")` returns the full weekday name, such as `Tuesday`.
- `current_time.strftime("%H:%M:%S")` uses the 24-hour clock.
- `current_time.strftime("%I:%M %p")` uses the 12-hour clock and adds `AM` or `PM`.

| Code | Meaning | Example |
|---|---|---|
| `%Y` / `%y` | Four-digit / two-digit year | `2026` / `26` |
| `%B` / `%b` | Full / abbreviated month | `November` / `Nov` |
| `%A` / `%a` | Full / abbreviated weekday | `Tuesday` / `Tue` |
| `%m` | Month number | `11` |
| `%d` | Day of the month, zero-padded | `10` |
| `%H` / `%I` | 24-hour / 12-hour clock hour | `21` / `09` |
| `%M` | Minute, zero-padded | `00` |
| `%S` | Second, zero-padded | `00` |
| `%p` | AM or PM marker | `PM` |

To display the actual current date and time instead of a fixed example, use `date.today()` and `datetime.now()`.

````python
# Fixed example values make the output predictable for learning.
today = date(2026, 11, 10)
current_time = datetime(2026, 11, 10, 21, 0, 0)

print(today.strftime("%d.%m.%Y"))
print(today.strftime("%B %d, %Y"))
print(today.strftime("%A"))
print(current_time.strftime("%H:%M:%S"))
print(current_time.strftime("%I:%M %p"))

# For live values, use:
# today = date.today()
# current_time = datetime.now()
````

### 11.2 Parse a date with `strptime()`

`strptime()` means **string parse time**: it converts a string into a date/datetime object using a format pattern.

Example requirement: input `2024_05_28` and display the date as `28.05.2024` (or `May 28, 2024`).

````python
from datetime import datetime

raw_date = "2024_05_28"
parsed_date = datetime.strptime(raw_date, "%Y_%m_%d")

print(parsed_date.strftime("%d.%m.%Y"))
print(parsed_date.strftime("%B %d, %Y"))
````

## 12. The `math` module

Python's built-in functions cover many common operations. The `math` module provides additional mathematical constants and functions. Import it before use.

````python
import math

print(math.pi)
print(math.e)
print(math.sqrt(81))
print(math.ceil(4.2))
print(math.floor(4.9))
print(math.factorial(5))
print(math.gcd(24, 36))
print(math.pow(2, 3))
````

### 12.1 Useful functions and constants

| Name | Purpose | Example |
|---|---|---|
| `math.sqrt(x)` | square root | `math.sqrt(25)` → `5.0` |
| `math.ceil(x)` | round up to an integer | `math.ceil(4.1)` → `5` |
| `math.floor(x)` | round down to an integer | `math.floor(4.9)` → `4` |
| `math.factorial(n)` | factorial of a non-negative integer | `math.factorial(5)` → `120` |
| `math.gcd(a, b)` | greatest common divisor | `math.gcd(12, 18)` → `6` |
| `math.pow(x, y)` | x raised to y, returned as float | `math.pow(2, 3)` → `8.0` |
| `math.pi` | π | circle calculations |
| `math.e` | Euler's number | exponential calculations |
| `math.radians(x)` | degrees to radians | `math.radians(180)` → π |
| `math.degrees(x)` | radians to degrees | `math.degrees(math.pi)` → `180.0` |
| `math.sin(x)`, `cos(x)`, `tan(x)` | trigonometric functions; input is in radians | `math.sin(math.pi / 2)` → `1.0` |

````python
import math

radius = 5
circumference = 2 * math.pi * radius
area = math.pi * radius ** 2

print(f"Circumference: {circumference:.2f}")
print(f"Area: {area:.2f}")

print(math.radians(180))
print(math.sin(math.pi / 2))
````

### Built-in functions vs. `math` functions

Do not confuse these similarly named operations:

- `round(4.7)` is a built-in function that rounds according to Python's rounding rules.
- `math.floor(4.7)` always rounds down to the lower integer.
- `math.ceil(4.1)` always rounds up to the higher integer.
- `abs(-8)` is built in; `math.fabs(-8)` returns a floating-point absolute value.
- `2 ** 3` is the exponentiation operator; `math.pow(2, 3)` returns a float.

````python
print(round(4.7))
print(math.floor(4.7))
print(math.ceil(4.1))
print(abs(-8), math.fabs(-8))
print(2 ** 3, math.pow(2, 3))
````

## 15. Quick reference

### String indexing and slicing

```python
text[index]
text[start:stop:step]
```

- `text[0]` — first character
- `text[-1]` — last character
- `text[:8]` — first eight characters
- `text[10:]` — from index 10 to the end
- `text[::3]` — every third character
- `text[::-1]` — reversed string
- `text[6:4:-1]` — move backwards from index 6, stopping before index 4

### Frequently used string methods

| Category | Methods | Purpose |
