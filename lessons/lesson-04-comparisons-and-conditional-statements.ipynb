## 2. Comparison Operators

Comparison operators compare two values. A comparison produces a Boolean value: `True` or `False`.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `<` | Less than | `3 < 5` | `True` |
| `>` | Greater than | `8 > 5` | `True` |
| `<=` | Less than or equal to | `5 <= 5` | `True` |
| `>=` | Greater than or equal to | `7 >= 5` | `True` |

### Assignment versus comparison

- `=` assigns a value to a variable.
- `==` checks whether two values are equal.

````python
age = 21
print(age == 21)
print(age == 18)
````

### Using comparison operators

````python
first_number = 10
second_number = 7

print(first_number == second_number)
print(first_number != second_number)
print(first_number < second_number)
print(first_number > second_number)
print(first_number <= 10)
print(second_number >= 10)
````

### Comparing strings

Strings can be compared too. String equality is case-sensitive, so `"Python"` and `"python"` are different strings.

````python
language = "Python"

print(language == "Python")
print(language == "python")
print(language != "Java")
````

Strings can also be compared with `<`, `>`, `<=`, and `>=` based on character ordering. For beginner programs, equality comparisons are especially common when checking whether text matches an expected value.

### Chained comparisons

Python allows comparisons to be chained. `18 <= age <= 65` checks that `age` is at least 18 and no more than 65. It is equivalent to using `and`.

````python
age = 25

print(18 <= age <= 65)
print(18 <= age and age <= 65)
````

## 3. Logical Operators

Logical operators combine conditions or reverse a Boolean value.

| Operator | Meaning | Example |
|---|---|---|
| `and` | True only if both sides are true | `True and False` → `False` |
| `or` | True if at least one side is true | `True or False` → `True` |
| `not` | Reverses a Boolean value | `not True` → `False` |

### 3.1 `and`

Both conditions must be true.

````python
age = 22
has_ticket = True

print(age >= 18 and has_ticket)
print(age >= 18 and age < 21)
````

### 3.2 `or`

At least one condition must be true.

````python
is_weekend = False
is_holiday = True

print(is_weekend or is_holiday)
print(is_weekend or False)
````

### 3.3 `not`

`not` reverses a Boolean result.

````python
is_logged_in = False

print(not is_logged_in)
print(not True)
print(not (10 > 5))
````

### Combining logical operators

Parentheses can make complex conditions easier to read and clarify the intended grouping.

````python
age = 24
has_membership = True
is_banned = False

can_access = (age >= 18 and has_membership) and not is_banned
print(can_access)
````

### Short-circuit evaluation

Python may skip evaluating the right side of a logical expression when the left side already determines the result.

- With `and`, a falsy left side makes the whole expression falsy.
- With `or`, a truthy left side makes the whole expression truthy.

This is called **short-circuit evaluation**.

````python
username = ""

if username and username == "admin":
    print("Welcome, administrator.")
else:
    print("A non-empty administrator name was not provided.")
````

## 4. Truthy and Falsy Values

Conditions do not have to contain only `True` or `False`. Python evaluates many values as **truthy** or **falsy**.

Common falsy values include:
- `False`
- `None`
- Numeric zero: `0`, `0.0`
- Empty strings: `""`
- Empty collections such as `[]`, `()`, and `{}`

Most other values are truthy, including non-empty strings and non-zero numbers. Use `bool()` to see the Boolean interpretation of a value.

````python
print(bool(0))
print(bool(15))
print(bool(0.0))
print(bool(-2.5))
print(bool(""))
print(bool("Python"))
print(bool(None))
print(bool("False"))
````

Notice that `bool("False")` is `True`: the string contains characters and is not empty. Python does not convert the text `"False"` into the Boolean value `False`.

### Using truthy/falsy values in a condition

A non-empty username can be checked directly:

````python
username = input("Enter a username: ")

if username:
    print(f"Hello, {username}!")
else:
    print("You did not enter a username.")
````

`if username:` checks whether the string is non-empty. This is a common Python style when you only need to check whether a value is present.

## 5. Conditional Statements: `if`, `elif`, and `else`

Conditional statements allow a program to choose which code block to execute.

### 5.1 `if`

An `if` block runs when its condition is truthy. Notice the colon and the indentation.

````python
age = 20

if age >= 18:
    print("You are an adult.")

print("Program finished.")
````

Indentation is part of Python syntax. Statements belonging to the same block must be indented consistently.

### 5.2 `else`

Use `else` to define what happens when the `if` condition is false.

````python
age = 16

if age >= 18:
    print("Access allowed.")
else:
    print("Access denied.")
````

### 5.3 `elif`

Use `elif` (“else if”) to check another condition when previous conditions were false. Python checks branches from top to bottom. The first matching branch runs.

````python
score = 82

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
else:
    print("Grade: F")
````

**Order matters.** Check the highest score range first. If `score >= 60` were checked first, a score of `95` would match that branch and later branches would be skipped.

### 5.4 Combining conditions

````python
age = 23
has_id = True

if age >= 18 and has_id:
    print("Entry allowed.")
else:
    print("Entry not allowed.")
````

### 5.5 Nested `if` statements

A nested `if` is an `if` statement inside another conditional block. It is useful when a second decision should be made only after the first condition is satisfied.

````python
is_registered = True
has_paid = True

if is_registered:
    print("Registration found.")
    if has_paid:
        print("Payment confirmed.")
        print("You can access the course.")
    else:
        print("Please complete your payment.")
else:
    print("Please register first.")
````

### Nested conditions versus combined conditions

A combined condition is suitable when checks belong together:

````python
age = 20
has_permission = True

if age >= 18 and has_permission:
    print("Access granted.")
else:
    print("Access denied.")
````

Nested conditions can provide a different response at each decision level:

````python
age = 20
has_permission = True

if age >= 18:
    if has_permission:
        print("Access granted.")
    else:
        print("Permission is required.")
else:
    print("You must be at least 18.")
````

Both approaches are valid. Use combined conditions for related checks. Use nested conditions when the program needs different actions or messages at each level.

## 6. Ternary Conditional Expression

A ternary conditional expression is a compact way to choose one of two values.

Syntax:

```python
value_if_true if condition else value_if_false
```

It is an expression, so it produces a value that can be assigned to a variable.

````python
age = 20
status = "Adult" if age >= 18 else "Minor"

print(status)
````

### Another example

````python
number = 14
result = "Even" if number % 2 == 0 else "Odd"

print(f"The number {number} is {result}.")
````

Use ternary expressions for simple two-way choices. For several branches or complex logic, a regular `if` / `elif` / `else` statement is usually easier to read.

## 7. `match` / `case`

The `match` statement compares a value against patterns and executes the first matching `case`.

It was introduced in **Python 3.10**. Use Python 3.10 or newer to run these examples.

Basic structure:

```python
match value:
    case pattern_1:
        # Code for pattern_1
    case pattern_2:
        # Code for pattern_2
    case _:
        # Default case
```

The underscore (`_`) is a wildcard pattern. It matches if no earlier case matches.

````python
command = "start"

match command:
    case "start":
        print("Starting the application...")
    case "stop":
        print("Stopping the application...")
    case "restart":
        print("Restarting the application...")
    case _:
        print("Unknown command.")
````

### Matching numbers

````python
day_number = 3

match day_number:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid day number.")
````

### Combining multiple cases

Use the `|` pattern to match more than one value in a single case.

````python
day = "Saturday"

match day:
    case "Saturday" | "Sunday":
        print("It is the weekend.")
    case "Monday" | "Tuesday" | "Wednesday" | "Thursday" | "Friday":
        print("It is a weekday.")
    case _:
        print("Unknown day.")
````

### `match` / `case` versus `if` / `elif`

Use `match` / `case` when comparing one value against several specific patterns. Use `if` / `elif` for ranges, inequalities, and general Boolean conditions. For example, checking whether a score is at least 90 is more naturally written with `if`.

## 8. Walrus Operator (`:=`)

The **walrus operator** (`:=`), also called the assignment expression operator, assigns a value to a variable as part of a larger expression. It was introduced in **Python 3.8**.

Regular assignment uses `=` as a standalone statement:

````python
message = "Hello, Python!"
print(message)
````

The walrus operator assigns a value inside an expression:

````python
if (length := len("Python")) > 5:
    print(f"The word contains {length} characters.")
````

In this example:
1. `len("Python")` calculates the string length.
2. `:=` assigns that result to `length`.
3. The condition checks whether `length` is greater than 5.
4. The indented block runs if the condition is true.

Parentheses make the assignment expression explicit and improve readability.

### Assigning input inside a condition

````python
if (username := input("Enter your username: ")):
    print(f"Welcome, {username}!")
else:
    print("The username cannot be empty.")
````

The input is read once and stored in `username`. The condition checks whether the entered string is non-empty.

### Checking a calculated value

````python
price = 120
discount = 15

if (final_price := price - (price * discount / 100)) < 100:
    print(f"The discounted price is {final_price:.2f}.")
else:
    print(f"The price after discount is {final_price:.2f}.")
````

### Important recommendations

- `:=` does not replace regular assignment (`=`).
- Use it when assigning and checking a value together makes the code clearer.
- Avoid it when it makes an expression difficult to understand.
- Parentheses are often required or recommended to make the intended expression clear.
- The assigned variable can be used after the expression within its applicable scope.

For beginners, ordinary assignment followed by a condition is often easier to read. Learn the walrus operator, but do not use it everywhere.

## 9. Putting the Concepts Together

### Example: Event Registration

This program combines input, type conversion, comparisons, logical operators, and conditional statements.

````python
age = int(input("Enter your age: "))
registration_code = input("Do you have a registration code? (yes/no): ")

if age >= 18 and registration_code == "yes":
    print("Registration successful.")
elif age >= 18 and registration_code == "no":
    print("You meet the age requirement, but need a registration code.")
elif age < 18:
    print("You must be at least 18 to register.")
else:
    print("Please enter a valid registration response.")
````

### Example: Simple Menu

This example uses `match` / `case` to select an action.

````python
print("1. View profile")
print("2. Edit profile")
print("3. Exit")

choice = input("Choose an option: ")

match choice:
    case "1":
        print("Opening profile...")
    case "2":
        print("Opening profile editor...")
    case "3":
        print("Goodbye!")
    case _:
        print("Invalid option.")
````
