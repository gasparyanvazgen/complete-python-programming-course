## 3. Python interpreter, script, and interactive mode

### Python interpreter

The Python interpreter reads and executes Python instructions. For now, remember that we use the interpreter to run Python code.

### Script

A **script** is a sequence of Python instructions saved in a file, usually with the `.py` extension. You can run the file whenever you need it without typing the code again. For example, `first_program.py` is a Python script.

### Interactive mode

In **interactive mode**, you type Python instructions directly into the Python environment and see the result immediately.

Open a terminal and run:

```bash
python
```

or:

```bash
python3
```

Then try this expression:

````python
2 + 3
````

The result should appear immediately. Now try:

````python
print("Hello")
````

To exit interactive mode, type `exit()`.

You can also use `Ctrl + D` on Linux/macOS. On Windows, use `Ctrl + Z`, then press `Enter`.

### Script vs. interactive mode

| Script | Interactive mode |
|---|---|
| Code is saved in a `.py` file. | Instructions are entered directly into the Python environment. |
| Useful for creating and saving programs. | Useful for quick experiments and testing small expressions. |
| You can run the whole file again. | You see the result after entering each instruction. |

## 4. The `print()` function: displaying text and numbers

`print()` is a built-in Python function. It displays the values you pass to it.

### Displaying text

Text values (strings) are written inside quotation marks.

````python
print("Hello, world!")
print('I am learning Python')
````

You can use either double quotation marks (`" "`) or single quotation marks (`' '`). Make sure the opening and closing quotation marks match.

### Displaying numbers

Numbers are usually written without quotation marks.

````python
print(10)
print(25 + 5)
print(10 * 3)
````

Notice the difference between these two statements:

````python
print(25 + 5)
print("25 + 5")
````

The first statement evaluates the expression and displays `30`. The second displays the text `25 + 5` exactly as written.

### Displaying multiple values

You can pass several values to `print()` by separating them with commas:

````python
print("Name:", "Alex")
print("Age:", 21)
print("Result:", 10 + 5)
````

## 5. Comments (`#`)

A **comment** is a part of the code that Python does not execute. Comments are written for people who read the code.

A single-line comment starts with the `#` symbol.

````python
# This is a comment
print("This line will run")

print("Hello")  # A comment can also appear after a statement
````

Use comments to explain why something is done when the reason is not obvious. Avoid writing comments that simply repeat what the code already says.

## 6. Syntax and indentation

**Syntax** means the rules for writing code in a programming language. If you break these rules, Python may report a `SyntaxError`.

For example, this code is incorrect because the closing parenthesis is missing:

```python
print("Hello"
```

Correct version:

````python
print("Hello")
````

**Indentation** means the spaces at the beginning of a line. Python uses indentation to define blocks of code.

We will study conditional statements and loops in later lessons. For now, remember that instructions belonging to the same block must be indented consistently.

In this example, the `print()` line is indented by four spaces:

````python
if True:
    print("This line is inside the block")
````

Use four spaces for each indentation level. Avoid mixing tabs and spaces.
