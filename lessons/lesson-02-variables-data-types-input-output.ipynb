## 2. Variables and Assignment

A **variable** is a name that refers to a value stored by a Python program.

Variables let us store information and use it later. For example, a program can store a person's name, age, or account balance.

### Creating a variable

Use the assignment operator `=` to assign a value to a variable.

````python
name = "David"
age = 21
height = 1.78

print(name)
print(age)
print(height)
````

In these examples:

- `name` refers to the string `"David"`.
- `age` refers to the integer `21`.
- `height` refers to the floating-point number `1.78`.
- `=` assigns the value on the right to the variable on the left.

**Important:** `=` means assignment in Python. It does not mean “is equal to” in the mathematical sense.

### Updating a variable

A variable can be assigned a new value. The latest assignment determines the value it refers to.

````python
score = 10
print(score)

score = 15
print(score)
````

### Assigning values to multiple variables

Python allows you to assign values to multiple variables in one statement.

````python
first_name, last_name, age = "Anna", "Smith", 20

print(first_name)
print(last_name)
print(age)
````

You can also assign the same value to multiple variables:

````python
x = y = z = 0

print(x, y, z)
````

### Variable naming rules

A variable name:

1. Must start with a letter (`a`–`z`, `A`–`Z`) or an underscore (`_`).
2. Cannot start with a number.
3. Can contain letters, numbers, and underscores.
4. Cannot contain spaces or special characters such as `-`, `@`, or `!`.
5. Cannot be a Python keyword such as `if`, `for`, or `class`.
6. Is case-sensitive: `age`, `Age`, and `AGE` are different names.

Examples:

````python
# Valid variable names
name = "Aram"
user_age = 21
score2 = 95
_private_value = 10

# Invalid examples (do not run these lines):
# 2score = 95
# user-name = "Aram"
# user age = 21
# class = "Python"
````

### Good naming practices

Choose descriptive names that explain what a value represents. Python commonly uses **snake_case** for variable names: words are lowercase and separated by underscores.

````python
# Less descriptive
x = 500
y = 12

# More descriptive
monthly_salary = 500
number_of_lessons = 12
````

## 3. Basic Data Types

A **data type** describes the kind of value a variable contains.

Python has several built-in data types. In this lesson, we will focus on five basic ones.

### 3.1 Integer — `int`

An integer is a whole number without a decimal part. It can be positive, negative, or zero.

````python
age = 21
temperature = -5
number_of_students = 30

print(age)
print(type(age))
````

### 3.2 Floating-point number — `float`

A `float` represents a number with a decimal part. Python uses a dot (`.`) as the decimal separator.

````python
price = 19.99
height = 1.78
temperature = -2.5

print(price)
print(type(price))
````

### 3.3 String — `str`

A string is a sequence of characters used to represent text. Strings are written inside single or double quotation marks.

````python
first_name = "Aram"
country = 'Armenia'
message = "I am learning Python."

print(first_name)
print(type(first_name))
````

A string can contain digits and still be a string:

````python
phone_number = "091123456"
number_as_text = "100"

print(type(phone_number))
print(type(number_as_text))
````

### 3.4 Boolean — `bool`

A Boolean value represents one of two logical states: `True` or `False`.

The first letter must be uppercase.

````python
is_student = True
has_job = False

print(is_student)
print(type(is_student))
````

Boolean values are often used to represent conditions or yes/no information. You will use them more when learning comparisons and conditional statements.

### 3.5 None — `NoneType`

`None` represents the absence of a value. It is useful when a variable has not received a meaningful value yet, or when a function intentionally returns no value.

The value is written as `None`, with an uppercase `N`.

````python
middle_name = None

print(middle_name)
print(type(middle_name))
````

### Checking a value's type with `type()`

Use `type(value)` to find the data type of a value.

````python
print(type(25))
print(type(3.14))
print(type("Python"))
print(type(True))
print(type(None))
````

### Compare the five types

| Data type | Example | Description |
|---|---|---|
| `int` | `42` | Whole number |
| `float` | `3.14` | Decimal number |
| `str` | `"Hello"` | Text |
| `bool` | `True` | Logical value |
| `NoneType` | `None` | Absence of a value |

## 4. Input and Output

Programs often need to receive information from users and display results.

- `input()` reads information entered by the user.
- `print()` displays information on the screen.

### 4.1 Reading input with `input()`

When Python executes `input()`, it displays an optional prompt and waits for the user to enter something.

````python
user_name = input("Enter your name: ")

print("Hello!")
print(user_name)
````

**Important:** `input()` always returns a string (`str`), even if the user enters digits.

Run this example and check the result:

````python
user_age = input("Enter your age: ")

print(user_age)
print(type(user_age))
````

If you enter `21`, Python still receives `"21"` as text. To perform numeric calculations, you usually need to convert the input.

### 4.2 Type conversion

Type conversion means changing a value from one data type to another.

Common conversion functions:

| Function | Converts to | Example |
|---|---|---|
| `int()` | Integer | `int("25")` → `25` |
| `float()` | Floating-point number | `float("3.5")` → `3.5` |
| `str()` | String | `str(25)` → `"25"` |
| `bool()` | Boolean | `bool(1)` → `True` |

````python
age_text = "21"
age = int(age_text)

price_text = "19.99"
price = float(price_text)

number = 100
number_text = str(number)

print(age, type(age))
print(price, type(price))
print(number_text, type(number_text))
````

### Converting user input

Here is a common pattern: read a value, convert it, and then use it in a calculation.

````python
age = int(input("Enter your age: "))
next_year_age = age + 1

print(next_year_age)
````

### Understanding `bool()` conversion

Python treats some values as **falsy** and most other values as **truthy** when converting them with `bool()`.

Common falsy values include:

- `False`
- `None`
- `0` and `0.0`
- `""` (an empty string)

Non-empty strings are truthy — even strings such as `"False"` and `"0"`.

````python
print(bool(0))
print(bool(1))
print(bool(""))
print(bool("Python"))
print(bool("False"))
print(bool(None))
````

**Important:** `bool("False")` is `True` because `"False"` is a non-empty string. Python does not interpret the text as the Boolean value `False`.

### Conversion errors

Not every string can be converted into a number. For example, `int("hello")` causes a `ValueError`.

For now, make sure the user enters a valid number when your program expects one. You will learn how to handle invalid input with exceptions in a later lesson.

````python
# Valid conversions
print(int("42"))
print(float("4.2"))

# This would cause a ValueError, so it is commented out:
# print(int("hello"))
````

## 5. Arithmetic Operators

Python supports the standard arithmetic operations.

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `8 + 3` | `11` |
| `-` | Subtraction | `8 - 3` | `5` |
| `*` | Multiplication | `8 * 3` | `24` |
| `/` | Division | `8 / 3` | `2.666...` |
| `//` | Floor division | `8 // 3` | `2` |
| `%` | Modulo (remainder) | `8 % 3` | `2` |
| `**` | Exponentiation | `2 ** 3` | `8` |

````python
a = 8
b = 3

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Exponentiation:", a ** b)
````

### Understanding `/`, `//`, and `%`

These three operators are especially important:

- `/` performs regular division and returns a `float`.
- `//` performs floor division: it rounds the quotient down to the nearest whole number.
- `%` returns the remainder after division.

For positive numbers, consider `17` divided by `5`:

````python
print(17 / 5)   # 3.4
print(17 // 5)  # 3
print(17 % 5)   # 2
````

The remainder operator is useful for checking whether a number is divisible by another number. For example, a number is even if its remainder after division by `2` is `0`.

````python
number = 14
remainder = number % 2

print(remainder)
print(remainder == 0)
````

### Operator precedence

When an expression contains multiple operators, Python follows rules that determine which operation is performed first.

A simplified order for the operators in this lesson is:

1. Parentheses: `()`
2. Exponentiation: `**`
3. Multiplication, division, floor division, and modulo: `*`, `/`, `//`, `%`
4. Addition and subtraction: `+`, `-`

Operators at the same level are generally evaluated from left to right, except exponentiation, which is evaluated from right to left.

````python
print(2 + 3 * 4)       # 14
print((2 + 3) * 4)     # 20
print(2 ** 3 ** 2)     # 512: 2 ** (3 ** 2)
print(20 / 5 * 2)      # 8.0: left to right
````

Use parentheses when they make an expression easier to understand. They also let you explicitly control the order of operations.

## 6. f-Strings: Formatting Text and Values

An **f-string** is a convenient way to insert variables or expressions into a string.

Put the letter `f` immediately before the opening quotation mark. Then place each variable or expression inside curly braces `{}`.

````python
name = "Aram"
age = 21

print(f"My name is {name}.")
print(f"I am {age} years old.")
````

### Inserting expressions

You can also put simple expressions inside the curly braces.

````python
a = 10
b = 5

print(f"{a} + {b} = {a + b}")
print(f"Twice {a} is {a * 2}.")
````

### Basic number formatting

You can control how numbers appear in an f-string.

For example, `:.2f` displays a floating-point number with two digits after the decimal point.

````python
price = 12.5
tax_rate = 0.2

print(f"Price: {price:.2f}")
print(f"Tax rate: {tax_rate:.0%}")
````

In the second example, `.0%` formats the value as a percentage with zero digits after the decimal point. The value `0.2` is displayed as `20%`.

More formatting options exist, but these are enough for basic output in this lesson.

## 7. Practical Work: Personal Information and Calculator

In this practical task, you will write a program that:

1. Asks the user for their name.
2. Asks for two numbers.
3. Converts the entered numbers into `float` values.
4. Calculates the sum, difference, product, and quotient.
5. Displays the results using f-strings.

### Step 1: Collect the user's information

Run the code and enter your own information.

````python
name = input("Enter your name: ")

first_number = float(input("Enter the first number: "))
second_number = float(input("Enter the second number: "))
````

### Step 2: Perform the calculations

Create variables for the calculation results.

````python
total = first_number + second_number
difference = first_number - second_number
product = first_number * second_number

print("Calculations completed.")
````

### Step 3: Display the results

Use f-strings to display readable output.

````python
print(f"Hello, {name}!")
print(f"{first_number} + {second_number} = {total}")
print(f"{first_number} - {second_number} = {difference}")
print(f"{first_number} * {second_number} = {product}")

if second_number != 0:
    quotient = first_number / second_number
    print(f"{first_number} / {second_number} = {quotient:.2f}")
else:
    print("Division is not possible because the second number is zero.")
````

The final example includes an `if` statement to prevent division by zero. Conditional statements will be studied in more detail in a later lesson. For now, focus on how the input, conversion, variables, arithmetic, and f-strings work together.

### Complete program

The following cell combines the steps into one program. Run it from top to bottom.

````python
name = input("Enter your name: ")

first_number = float(input("Enter the first number: "))
second_number = float(input("Enter the second number: "))

total = first_number + second_number
difference = first_number - second_number
product = first_number * second_number

print()
print(f"Hello, {name}!")
print(f"Sum: {total:.2f}")
print(f"Difference: {difference:.2f}")
print(f"Product: {product:.2f}")

if second_number != 0:
    quotient = first_number / second_number
    print(f"Quotient: {quotient:.2f}")
else:
    print("Quotient: cannot divide by zero.")
````
