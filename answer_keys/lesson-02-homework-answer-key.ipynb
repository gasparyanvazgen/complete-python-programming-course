### Exercise 2: Type table

**Idea:** The surprise is `10 / 5`. The answer is `2.0`, a `float`. The `/` operator always gives a float. `7 > 3` gives a `bool`.

````python
print(type(7))        # int
print(type(7.0))      # float
print(type("7"))      # str
print(type(7 > 3))    # bool
print(type(None))     # NoneType
print(type(10 / 5))   # float (/ always gives a float: 2.0)
print(type(10 // 5))  # int
print(type(2 ** 3))   # int
````

### Exercise 3: True or False?

**Idea:** These values become `False`: `False`, `None`, `0`, `0.0` and `""`. Almost everything else becomes `True`. A single space `" "` is not empty.

````python
print(bool(0.0))    # False: zero is False
print(bool("0"))    # True: the text is not empty
print(bool(""))     # False: empty text
print(bool(" "))    # True: a space is a character, so it is not empty
print(bool(-1))     # True: any number except 0 is True
print(bool(None))   # False

# bool("False") is True because "False" is text and the text is not empty.
# Python does not read the meaning of the text. It only checks if the text is empty.
print(bool("False"))
````


### Exercise 4: Follow the variables

**Idea:** `b = a` copies the value. If `a` changes later, `b` stays the same. The last line gives `c` and `d` the same value: `13 // 3 = 4`.

````python
a = 5
b = a
print(a, b)        # 5 5   (b gets a copy of 5)
a = 8
print(a, b)        # 8 5   (b does not change when a changes)
b = b + a
print(a, b)        # 8 13  (5 + 8)
c = d = b // 3
print(a, b, c, d)  # 8 13 4 4  (13 // 3 = 4)
````

**Sample run:**

```text
5 5
8 5
8 13
8 13 4 4
```

### Exercise 7: Swap two values

**Idea:** Way 1: save one value before you change it. Way 2: keep the sum in `a`, then use subtraction to get each old value back.

````python
a = 4
b = 9

# Way 1: third variable
temp = a
a = b
b = temp
print(a, b)  # 9 4

# Way 2: only + and -
a = 4
b = 9
a = a + b    # a = 13
b = a - b    # b = 13 - 9 = 4
a = a - b    # a = 13 - 4 = 9
print(a, b)  # 9 4

# Extra: Python can swap in one line
a = 4
b = 9
a, b = b, a
print(a, b)  # 9 4
````

### Exercise 9: Seconds to clock

**Idea:** Use two steps. First take the hours (1 hour = `3600` seconds). Then split what is left into minutes and seconds.

### Exercise 10: Digits

**Idea:** `% 10` gives the last digit. `// 10` removes it. The first digit is `number // 100`. The middle digit is the last digit of `number // 10`.

````python
number = int(input("Enter a three-digit number: "))

last_digit = number % 10
without_last = number // 10
digit_sum = number // 100 + (number // 10) % 10 + number % 10

print(f"Last digit: {last_digit}")
print(f"Without the last digit: {without_last}")
print(f"Sum of digits: {digit_sum}")
````

### Exercise 13: Rectangle

**Idea:** The perimeter needs parentheses. `2 * width + height` is wrong because `*` goes before `+`.

````python
width = float(input("Enter width: "))
height = float(input("Enter height: "))

area = width * height
perimeter = 2 * (width + height)
diagonal_squared = width ** 2 + height ** 2

print(f"Area: {area:.2f}")
print(f"Perimeter: {perimeter:.2f}")
print(f"Diagonal squared: {diagonal_squared:.2f}")
````

### Exercise 16: Average

**Idea:** This is about order of operations. Parentheses make the addition happen first, before the division.

````python
score_1 = float(input("Score 1: "))
score_2 = float(input("Score 2: "))
score_3 = float(input("Score 3: "))

average = (score_1 + score_2 + score_3) / 3
print(f"Average: {average:.2f}")

# Without parentheses, only score_3 is divided by 3,
# because division goes before addition. So the answer is wrong.
wrong = score_1 + score_2 + score_3 / 3
print(f"Wrong answer: {wrong:.2f}")
````

### Exercise 20: Safe calculator

**Idea:** The three operations can all fail when the second number is 0, so they are in the same `if` block. `**` is safe, so it stays outside.

````python
first = int(input("First number: "))
second = int(input("Second number: "))

if second != 0:
    print(f"{first} / {second} = {first / second:.2f}")
    print(f"{first} // {second} = {first // second}")
    print(f"{first} % {second} = {first % second}")
else:
    print("You cannot use /, // or % because the second number is zero.")

print(f"{first} ** {second} = {first ** second}")

if second != 0:
    print(f"Can {first} be divided by {second} with no remainder? {first % second == 0}")
````

**Sample run (typed: 17, 5):**

```text

Second number: 5

17 // 5 = 3
17 % 5 = 2
17 ** 5 = 1419857
Can 17 be divided by 5 with no remainder? False
```
