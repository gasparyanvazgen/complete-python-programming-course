## Exercise 05: Find a Word in a Sentence

**Idea:** `find()` gives `-1` when the word is not found. `index()` gives an error when the word is not found. So check first with `in`, and call `index()` only if the word exists.

````python
text = "Learning Python makes problem solving enjoyable"
word = input("Word to search for: ")

if word in text:
    print("The word is in the sentence: True")
    print("find() index:", text.find(word))
    print("index() index:", text.index(word))
else:
    print("The word is in the sentence: False")
    print("find() index:", text.find(word))
    print("The word was not found, so index() is not used.")
````

## Exercise 09: Replace Sensitive Information

**Idea:** `replace(old, new)` changes every match. The third argument is the maximum number of changes. Use `1` to change only the first match.

## Exercise 12: Product Price Calculator

**Idea:** Discount amount = price * percent / 100. `{name:^25}` centers the text in 25 characters. `:.1%` shows a fraction as a percent, so `0.15` becomes `15.0%`.

````python
product_name = input("Product name: ")
original_price = float(input("Original price: "))
discount_percent = float(input("Discount percent: "))

discount_rate = discount_percent / 100
discount_amount = original_price * discount_rate
final_price = original_price - discount_amount

print(f"[{product_name:^25}]")
print(f"Original price:  {original_price:.2f}")
print(f"Discount rate:   {discount_rate:.1%}")
print(f"Discount amount: {discount_amount:.2f}")
print(f"Final price:     {final_price:.2f}")
````

## Exercise 14: Check Data Types with isinstance()

**Idea:** `type(x)` shows the type. `isinstance(x, int)` checks the type and gives `True` or `False`. A tuple like `(int, float)` means "one of these types". Note: `bool` is also a kind of `int` in Python, so `isinstance(True, int)` is `True`.

````python
age = 21
price = 19.99
student_name = "Anna"
is_enrolled = True

print("age:", type(age), isinstance(age, int))
print("price:", type(price), isinstance(price, float))
print("student_name:", type(student_name), isinstance(student_name, str))
print("is_enrolled:", type(is_enrolled), isinstance(is_enrolled, bool))

print("price is int or float:", isinstance(price, (int, float)))
````

## Exercise 16: Safe Expression Awareness

**Idea:** Use `if / elif / else` and choose the operation by the symbol. Never use `eval()`. Check for zero before division.

````python
number_1 = float(input("First number: "))
number_2 = float(input("Second number: "))
operation = input("Operation (+, -, *, /): ").strip()

if operation == "+":
    print(f"Result: {number_1 + number_2:.2f}")
elif operation == "-":
    print(f"Result: {number_1 - number_2:.2f}")
elif operation == "*":
    print(f"Result: {number_1 * number_2:.2f}")
elif operation == "/":
    if number_2 == 0:
        print("Error: you cannot divide by zero.")
    else:
        print(f"Result: {number_1 / number_2:.2f}")
else:
    print("Error: use only +, -, * or /.")
````

## Exercise 17: Date String Converter

**Idea:** `strptime(text, format)` reads a date from text. Format codes: `%Y` year, `%m` month, `%d` day. For output: `%B` month name, `%A` weekday name. Put the code in `try` / `except ValueError` to catch a wrong date.

````python
from datetime import datetime

text = input("Enter a date (YYYY_MM_DD): ").strip()

try:
    date = datetime.strptime(text, "%Y_%m_%d")
    print("Format 1:", date.strftime("%d.%m.%Y"))
    print("Format 2:", date.strftime("%B %d, %Y"))
    print("Weekday:", date.strftime("%A"))
except ValueError:
    print("Invalid date or wrong format. Use YYYY_MM_DD, for example 2024_05_28.")
````

## Exercise 18: Date Difference Calculator

**Idea:** Two `date` objects can be subtracted. The result has `.days`. Use `abs()` to get a positive number of days.

````python
from datetime import date

text_1 = input("First date (YYYY-MM-DD): ").strip()
text_2 = input("Second date (YYYY-MM-DD): ").strip()

try:
    date_1 = date.fromisoformat(text_1)
    date_2 = date.fromisoformat(text_2)

    days = abs((date_2 - date_1).days)
    print("Difference in days:", days)

    if date_1 < date_2:
        print("The earlier date is:", date_1)
    elif date_2 < date_1:
        print("The earlier date is:", date_2)
    else:
        print("The dates are the same.")
except ValueError:
    print("Invalid date. Use YYYY-MM-DD, for example 2024-05-28.")
````

## Exercise 19: Circle and Geometry Calculator

**Idea:** Area = `math.pi * r ** 2`. Circumference = `2 * math.pi * r`. The diameter is `2 * r`. The diagonal of a square with side `s` is `math.sqrt(s ** 2 + s ** 2)`. Check for a negative radius first.

````python
import math

radius = float(input("Enter the radius: "))

if radius < 0:
    print("The radius cannot be negative. Please enter a number that is 0 or bigger.")
else:
    area = math.pi * radius ** 2
    circumference = 2 * math.pi * radius
    diameter = 2 * radius
    diagonal = math.sqrt(diameter ** 2 + diameter ** 2)

    print(f"Area: {area:.2f}")
    print(f"Circumference: {circumference:.2f}")
    print(f"Diagonal of the square (side = diameter): {diagonal:.2f}")
````

## Exercise 20: Math Library Challenge

**Idea:** Check the input first with `isdigit()` and `int(...) > 0`. Then use the `math` functions. `math.radians()` changes degrees to radians, so here the number is the degrees.

````python
import math

text = input("Enter a positive whole number: ").strip()

if text.isdigit() and int(text) > 0:
    number = int(text)
    root = math.sqrt(number)

    print(f"Square root: {root:.4f}")
    print("Factorial:", math.factorial(number))
    print("Ceiling of the square root:", math.ceil(root))
    print("Floor of the square root:", math.floor(root))
    print(f"{number} degrees in radians: {math.radians(number):.4f}")
else:
    print("Invalid input. Please enter a positive whole number, like 10.")
````
